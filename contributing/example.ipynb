## Displaying a chart

The cell below generates a chart. Note that the cell has been labelled using `#| label: sitting-days-chart` this means the cell can be referenced elsewhere using `sitting-days-chart`.

If you look in the cell metadata you'll see it includes the `remove-cell` tag. This means the cell and its output won't be included in the final book. Instead we'll embed the chart in a figure below so we can add a caption and it will be automatically numbered.

````python
#| label: sitting-days-chart

import pandas as pd
import altair as alt

df = pd.read_json("https://handbookapi.aph.gov.au/api/StatisticalInformation/SittingDaysPerYear")

dfs = []
for house in ["Reps", "Senate"]:
    df_h = df[["Year", f"{house}Days"]]
    df_h["house"] = house
    df_h.columns = ["year", "days", "house"]
    dfs.append(df_h)
df_split = pd.concat(dfs)

ticks = [t for t in range(1900, 2030, 5)]
alt.Chart(df_split).mark_bar().encode(
    x=alt.X("year:O").axis(values=ticks).title(None),
    y=alt.Y("days:Q").title(None),
    row=alt.Row("house").title("number of days"),
    color=alt.Color("house").legend(None),
    tooltip=["year:O", "house:N", "days:Q"]
).properties(height=200, width=600, padding=20)
````

Now we'll reference the `#sitting-days-chart` in a `figure` directive and give it a caption. We'll also label the figure so that it can be referenced elsewhere.

```Myst
:::{figure} #sitting-days-chart
:label: sitting-days-chart-fig

Number of sitting days per year for the House of Representatives and the Senate
:::
```
