# Guide for authors

## Front matter

Include extra metadata in the front matter section of a notebook. Possible values include:

- `authors`
- `abstract`
- `keywords`

For example, to list the notebook's authors add their details to the **first cell** of the notebook like this.

```yaml
---
authors:
    - name: Tim Sherratt
      orcid: http://orcid.org/0000-0001-7956-4498
      url: https://timsherratt.au
      mastodon: "@wragge@hcommons.social"
      github: "@wragge"
---
```

You can also include author information in the `project` configuration settings in the `myst.yml` file. If you include an `id` value (such as 'tim') in the configuration you can use this in the notebook frontmatter instead of repeating all the details.

```yaml
---
authors:
    - tim
---
```

Front matter rules:
- has to be in the first cell of the notebook
- cell has to be set to 'Markdown'
- data has to be valid YAML
- YAML data has to be enclosed top and bottom by `---` lines

If your front matter cell is rendering as plain text, check your YAML for errors.

## Labelling cells and content

By labelling code and Markdown cells they can used in cross-references or embedded elsewhere in the book.

Note that labels are accessible across the whole book project, not just a notebook, so should be unique for the book. Make them descriptive, eg `reps-words-per-year-chart` rather than `chart-1`.

To label a code cell, add `#| label: my-cell-label` at the top of the cell, eg: 

```#| label: reps-words-per-year-chart```

To label a Markdown block put `(my-label)=` on the line above it, for example:

```
(labelling-example)=
This is an example of labelling markdown
```

Labels can also be added to directives.

```
:::{figure} #my-chart
:label: my-chart-figure

This is my chart.
:::
```

## Links and cross-references

:::{warning}
Note that the Myst extension for Jupyter Lab doesn't seem to render some embeds and cross references from other notebooks. They are, however, rendered correctly when the `jupyter book build` command is run. So even if it looks broken in the notebook, it might be ok in the site once it's built.
:::


To create an internal link to another notebook, use standard Markdown syntax with the full path to the notebook in the file structure, with or without the file extension. If you leave the square brackets empty, the title of the notebook will be used as the link text.

For example, `[](/hansard/hansard-intro)` produces:

[](/hansard/hansard-intro)

To create an internal link to a heading in another notebook, use standard Markdown syntax with the full path to the notebook and the heading's HTML `id`. If you leave the square brackets empty, the text of the heading will be used as the link text. A preview of the section will be displayed when you hover over the link.

For example to link to the 'Jupyter Book' heading in the 'contributing' notebook, use `[](/contributing/contributing#jupyter-book)`:

[](/contributing/contributing#jupyter-book)

To create an internal link to a component that you've added a label to, use standard Markdown syntax with the label name preceded by a hash symbol `#`, eg `[](#sitting-days-chart-fig)`:

Here's a [chart showing the number of sitting days per year](#sitting-days-chart-fig) in the example notebook.

You can also use the `@sitting-days-chart-fig` shorthand:

Here's @sitting-days-chart-fig

And the `{ref}` role:

Here's {ref}`sitting-days-chart-fig`

## Code cell tags

:::{csv-table}
:header: Tag,	Description
`remove-cell`, Remove the cell from the rendered output
`remove-input`,	Remove the code cell input/source from the rendered output.
`remove-output`, Remove the code cell output from the rendered output.
`hide-cell`,	Hides the cell behind a dropdown in the rendered output.
`hide-input`,	Hides the code cell input/source behind a dropdown in the rendered output.
`hide-output`,	Hides the code cell output behind a dropdown in the rendered output.
:::

## Style

Follow the *Australian Government Style Manual* in naming government entities:

- [Government terms](https://www.stylemanual.gov.au/grammar-punctuation-and-conventions/names-and-terms/government-terms)
- [Parliaments and Councils](https://www.stylemanual.gov.au/grammar-punctuation-and-conventions/titles-honours-forms-address/parliaments-and-councils)
