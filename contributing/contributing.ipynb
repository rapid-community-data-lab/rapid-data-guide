## Jupyter Book

This site is built using [Jupyter Book](https://jupyterbook.org/). Jupyter Book takes a collection of Markdown files and Jupyter notebooks and renders them as a static HTML  site with a book-like structure. The Jupyter notebooks can include code cells that are executed at build time to generate tables, visualisations and other outputs.



## Local installation

To set up a local development environment:

1. git clone the repository
2. create a Python virtual environment
3. install the requirements
4. start Jupyter Book
5. create/edit notebooks in Jupyter Lab
