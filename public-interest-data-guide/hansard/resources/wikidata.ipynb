## Legislation

Query to retrieve Australian federal legislation with dates of assent and repeal and any associated ComLaw identifiers. [Try it >](https://w.wiki/U9Lp)

```sparql
SELECT ?item ?itemLabel ?assent ?repeal (GROUP_CONCAT(?comlaw; separator = " | ") as ?comlaw_ids)
WHERE
{
  ?item wdt:P31 wd:Q56731284.
  OPTIONAL {
    ?item p:P793 ?date1.
    ?date1 ps:P793 wd:Q1070654.
    ?date1 pq:P585 ?assent}
  OPTIONAL {
    ?item p:P793 ?date2.
    ?date2 ps:P793 wd:Q4590027.
    ?date2 pq:P585 ?repeal}
  OPTIONAL {
    ?item wdt:P2461 ?comlaw.
  }
  SERVICE wikibase:label { bd:serviceParam wikibase:language "[AUTO_LANGUAGE],mul,en". } # Helps get the label in your language, if not, then default for all languages, then en language
}
GROUP BY ?item ?itemLabel ?assent ?repeal
ORDER BY ASC(?assent)

```
