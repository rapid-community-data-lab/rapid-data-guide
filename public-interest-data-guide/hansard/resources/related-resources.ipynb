## Other Hansard resources

### Historic Hansard

[Commonwealth of Australia parliamentary debates from 1901 to 1980](https://historichansard.net) presented in an easy-to-read format for historians and other lovers of political speech.

### Datasets

- [hansard-xml](https://github.com/wragge/hansard-xml) – repository containing Hansard XML (and some SGML) files from 1901 to 2005
- [A new, comprehensive database of all proceedings of the Australian Parliamentary Debates (1998-2025)](https://doi.org/10.5281/zenodo.7336075)
- [federal-hansard-db](https://github.com/fonzzy1/federal-hansard-db) – a structured PostgreSQL database of the Australian Federal Parliament’s Hansard
- [hansard-interjections](https://github.com/wragge/hansard-interjections) – a collection of 981,706 interjections were extracted from Hansard XML files, 1901 to 1980

## Parliament

### Parliamentary Handbook

The [*Parliamentary Handbook of the Commonwealth of Australia*](https://handbook.aph.gov.au) provides detailed information on Parliament, parliamentarians, political parties, elections and electorates. It is a critical resource for understanding the context of Hansard. The *Handbook* is built on top of [an API](/hansard/resources/parliamentary-handbook-api.ipynb) that can be used to extract useful data in machine-readable form.

The Handbook and Hansard use the same identifiers for parliamentarians. This makes it easy to link speeches in Hansard to the biographical details of individual speakers. [Provide some examples.]{.mark}

### ParlInfo

[ParlInfo](https://parlinfo.aph.gov.au/parlInfo/search/search.w3p) is a massive database of material documenting the history, context, and operation of the Australian Parliament. As well as Hansard, it includes a wide variety of publications such as proceedings, reports, and parliamentary papers. However, it is so large and diverse that it can be hard to navigate.

### Parliamentary proceedings

Hansard is complemented by the official record of the decisions and proceedings of each house of Parliament – the *Votes and Proceedings* of the House of Representatives, and the *Journal* of the Senate. These are available in ParlInfo, and also via the following pages:

- [Votes and Proceedings of the House of Representations, 1901-2025](https://www.aph.gov.au/Parliamentary_Business/Chamber_documents/HoR/Votes_and_Proceedings/votes1p3p)
- [Journals of the Senate, 1901-2002](https://www.aph.gov.au/Parliamentary_Business/Chamber_documents/Senate_chamber_documents/Journals_of_the_Senate/Pre_2002_Journals)
- [Journals of the Senate, 2002-2026](https://www.aph.gov.au/Parliamentary_Business/Chamber_documents/Senate_chamber_documents/Journals_of_the_Senate)

### Parliamentary papers

Parliamentary papers, or tabled papers, are documents presented to the Australian Parliament. Sometimes this is required by law. Other times it's just for information. As the [Parliament of Australia website explains](https://www.aph.gov.au/Parliamentary_Business/Chamber_documents/Tabled_Papers):

> Documents tabled in Parliament include reports from parliamentary committees, annual reports of departments and agencies, reports from the Auditor-General and legislative instruments.

Parliamentary papers can be found in [ParlInfo](https://parlinfo.aph.gov.au/parlInfo/search/summary/summary.w3p;adv=yes;orderBy=customrank;page=0;query=Dataset%3Atabledpapers;resCount=Default) and Trove. The *Trove Data Guide* provides an [overview](https://tdg.glam-workbench.net/other-digitised-resources/parliamentary-papers/overview.html) of the digitised parliamentary papers through Trove, as well as some [hints for finding them](https://tdg.glam-workbench.net/other-digitised-resources/parliamentary-papers/finding-pp.html).


![Dates of Parliamentary Papers in Trove ([*Trove Data Guide*](https://tdg.glam-workbench.net/other-digitised-resources/parliamentary-papers/overview.html#pp-overview-chart-dates))](/images/pp-papers-trove.svg){#fig-trove-ppapers}

### Government Gazette

The Australian Government is required to publish information about changes to legislation in the *Government Gazette*. Copies of the Government Gazette are available in the [Federal Register of Legislation](https://www.legislation.gov.au/gazettes), and [Trove](https://trove.nla.gov.au/newspaper/?type=gazette).

## Legislation

### Federal Register of Legislation

The Federal Register of Legislation provides an official list of the Acts of the Australian Parliament, together with related documents such as rules and regulations. As of September 2026, [13,732 Acts were listed](https://www.legislation.gov.au/search/isasmade()/pointintime(1901-01-01,2026-09-11)/collection(Act)/sort(name%20asc)). Data from the Register is available [through an API]( [API](https://www.legislation.gov.au/help-and-resources/using-the-legislation-register/data-share-and-reuse)! ).

Acts start their lives as bills, presented to Parliament for the consideration of members. Most of the discussion in Hansard relates to bills. Once passed by Parliament, bills gain Royal Assent and become Acts of Parliament. Over time, Acts can be amended or repealed by other Acts, creating complex histories. All of these changes and relationships are captured by Federal Register, and each version of an Act is assigned its own identifier.

Of course, not all bills are passed. You can [search ParlInfo for details of bills](https://parlinfo.aph.gov.au/parlInfo/search/summary/summary.w3p;adv=yes;orderBy=customrank;page=0;query=Dataset%3AbillsPrevParl;resCount=Default) presented to Parliament.

### AustLII (Australasian Legal Information Institute)

[AustLII](https://www.austlii.edu.au/) provides a large collection of databases documenting Australian legislation and court law.

## Sources of political speech

### PM Transcripts

A [collection of 20,000+ transcripts](https://pmtranscripts.pmc.gov.au) of speeches, interviews, and press releases delivered by Australian Prime Ministers from the 1940s to the present. The transcripts are also available as XML files via a [simple API](https://pmtranscripts.pmc.gov.au/developers).

![Number of transcripts by Prime Minister and year](/images/pm-transcripts.svg){#fig-pm-transcripts .column-body-outset}

The [GLAM Workbench](https://www.glam-workbench.net/pm-transcripts/) includes examples of harvesting, indexing, aggregating and visualising the transcripts. The full collection of XML files, as well as a CSV index, and aggregations by Prime Minister, are available in [this GitHub repository](https://github.com/wragge/pm-transcripts-data).

### Election Speeches (Museum of Australian Democracy)

[A recent update to [the site](https://moadoph.gov.au/explore/democracy/election-speeches) removed the link to download the speeches as a dataset.]{.mark}

### Press release database (Parliamentary Library)

ParlInfo includes a [collection of thousands of press releases, speeches, and interview transcripts](https://parlinfo.aph.gov.au/parlInfo/search/summary/summary.w3p;adv=yes;orderBy=customrank;page=0;query=Dataset%3Apressrel;resCount=Default) issued by Australian federal politicians and saved by the Parliamentary Library. More than [380,000 of these press releases](https://trove.nla.gov.au/search/category/books?keyword=nuc%3A%22APAR%3APR%22) have been indexed in Trove.

The GLAM Workbench includes examples of [harvesting press releases from Trove and ParlInfo](https://www.glam-workbench.net/trove-government/harvest-parliament-press-releases/). Sample datasets of press releases relating to [COVID](https://www.glam-workbench.net/trove-government/trove-parliament-press-releases-covid/) and [refugees](https://www.glam-workbench.net/trove-government/trove-parliament-press-releases-refugees/) are available for download.
