---

This site was created by the [Reusable and Accessible Public Interest Documents (RAPID-CDL) project](https://rapid-cdl.edu.au), part of [ARDC’s Community Data Lab](https://ardc.edu.au/project/ardc-community-data-lab/). The RAPID-CDL project aims to make public interest documents such as parliamentary proceedings, inquiry submissions, hearing transcripts, and ministerial records more accessible and usable for researchers across Australia.
