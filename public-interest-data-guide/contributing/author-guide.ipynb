## Front matter

Extra metadata can be added in the front matter section of a notebook. At a minimum, provide a value for `authors`. While you're developing a new notebook set `draft` to `true` – this will add a 'draft' header in the development version and exclude the notebook from the production version (see [more on dev and prod sites](/contributing/contributing.ipynb#sec-dev-prod-sites)).

For example, to list the notebook's authors add their details to the **first cell** of the notebook like this.

```yaml
---
authors:
    - name: Tim Sherratt
      orcid: http://orcid.org/0000-0001-7956-4498
      url: https://timsherratt.au
draft: true
---
```

If you add these fields, they'll be automatically added to the page header

- `abstract`
- `keywords`

Front matter rules:

- has to be in the first cell of the notebook
- cell has to be set to 'Markdown'
- data has to be valid YAML
- YAML data has to be enclosed top and bottom by `---` lines

## Labelling cells and content

By labelling code and Markdown cells they can used in cross-references or embedded elsewhere in the book.

Note that labels are accessible across the whole site, not just a notebook, so should be unique. Quarto defines a set of prefixes that should be used in labels, eg: 'fig', 'tbl', and 'sec' for figures, tables and sections.

To label a [section](https://quarto.org/docs/authoring/cross-references.html#sections), add `{#sec-my-section-name}` after a heading. For example:

```markdown
## Cool things about Hansard {#sec-my-cool-hansard-things}
```

To label an [image](https://quarto.org/docs/authoring/cross-references.html#overview) add `{#fig-my-image}` after the standard Markdown image embed link:

```markdown
![](parliament-house.jpeg) {#fig-parliament-house}
```

To label an [image with a caption](https://quarto.org/docs/authoring/cross-references.html#figures):

```markdown
::: {#fig-parliament-house}

![](parliament-house.jpeg)

Parliament House in Canberra
:::

```

To label a code cell that produces a chart add `#| label: fig-my-chart` and `#| fig-caption: My interesting chart` to the very top of the cell:

```python
#| label: fig-sitting-days
#| fig-cap: Number of sitting days per year for the House of Representatives and the Senate

# Code goes here
```

## Links and cross-references

To create an internal link to another notebook, use standard Markdown syntax with the full or relative path to the notebook in the file structure, including the file extension. 

For example, `[introduction to Hansard](/hansard/hansard-intro.ipynb)` produces:

> Here's an [introduction to Hansard](/hansard/hansard-intro.ipynb)

To create an internal link to a heading in another notebook, use standard Markdown syntax with the full path to the notebook and the heading's HTML `id`. For example to link to the 'Quarto' heading in the 'contributing' notebook, use `[](/contributing/contributing#quarto)`:

> This site uses [quarto](/contributing/contributing.ipynb#quarto).

## Code rendering/execution

Code cells will be displayed by default. To hide all the code cells in a notebook behind a 'Code' dropdown, add the following to the front matter:

```
format:
  html:
    code-fold: true
```

To completely hide a particular cell, add this to the top of the code cell: `#| echo: false`.

To include a variable defined in a code cell with Markdown text, first add the following to the front matter:

```
execute:
  enabled: true
```
Then use \``{{python}} my-variable`\` in your Markdown.

````python
import datetime
date = datetime.datetime.now().isoformat()
````

Markdown code: The date is \``{{python}} date`\`

The date is `{python} date`
